```python
import pandas as pd
import os

csv_path = 'C:\\Users\\subha\\OneDrive\\Documents\\GitHub\\solar-agro-irrigation-system\\data\\processed\\processed_dataset_Created.csv'
if os.path.exists(csv_path):
	df = pd.read_csv(csv_path)
	print(df.columns)
else:
	print(f"File not found: {csv_path}")
```

```python
%pip install scikit-learn
%pip install joblib

import joblib

model = joblib.load("C:\\Users\\subha\\OneDrive\\Documents\\GitHub\\solar-agro-irrigation-system\\models\\moisture_forecast_maharashtra.pkl")

# Select only features used by the forecast model
FEATURES = [
    'Aggregate Soilmoisture Percentage (at 15cm)',
    'Aggregate Soilmoisture Percentage (at 15cm)_lag_1',
    'Aggregate Soilmoisture Percentage (at 15cm)_lag_3',
    'Aggregate Soilmoisture Percentage (at 15cm)_lag_7',
    'Aggregate Soilmoisture Percentage (at 15cm)_roll7',
    'month_sin','month_cos',
    'sandy','floamy','clayey','clayskeletal','drainage_score'
]

# Drop rows with NaNs in the features before prediction
df_no_nan = df.dropna(subset=FEATURES)
df.loc[df_no_nan.index, 'pred_moisture_t1'] = model.predict(df_no_nan[FEATURES])[:, 0]  # take t+1 prediction
df.to_csv(csv_path, index=False)
print("✅ Added pred_moisture_t1 to processed CSV")
```
